### Binary Search

```python
def binary_search(arr, target):
    """
    Perform binary search on a sorted array to find the index of the target value.

    Parameters:
    arr (list): A sorted list of elements.
    target: The value to search for in the array.

    Returns:
    int: The index of the target value if found, otherwise -1.
    """
    low = 0
    high = len(arr) - 1

    # The while loop drives the O(log n) splits
    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid  # Target found!
        elif arr[mid] < target:
            low = mid + 1  # Discard the left half
        else:
            high = mid - 1  # Discard the right half

    return -1  # Target not found

#Time complexity of binary search is O(log n) where n is the number of elements in the array.
```

```python
binary_search([1, 2, 3, 4, 5, 6, 7, 8, 9], 9)
```

### Selection Sort

```python
def selection_sort(arr):
    """
    Perform selection sort on an array.

    Parameters:
    arr (list): A list of elements to be sorted.

    Returns:
    list: The sorted list.
    """
    n = len(arr)
    for i in range(n):
        # Assume the minimum is the first element
        min_index = i
        for j in range(i + 1, n):   
            if arr[j] < arr[min_index]:
                min_index = j  # Update min_index if a smaller element is found
        # Swap the found minimum element with the first element
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr
```

```python
selection_sort([9, 1, 4, 9, 5])
```

### Insertion Sort

```python
def insertion_sort(arr):
    """
    Perform insertion sort on an array.

    Parameters:
    arr (list): A list of elements to be sorted.

    Returns:
    list: The sorted list.
    """
    for i in range(1, len(arr)):
        j = i
        while (j > 0 and arr[j] < arr[j-1]):
            arr[j], arr[j-1] = arr[j-1], arr[j]
            j -= 1
    return arr
```

```python
insertion_sort([5, 4, 3, 2, 1])
```

### Merge Sort

```python
def merge(A, B):
    (m, n) = len(A), len(B)
    (C, i, j) = ([], 0 , 0)
    
    # Case 1 :- When both lists A and B have elements for comparing
    while i < m and j < n:
        if A[i] <= B[j]:
            C.append(A[i])
            i += 1
        else:
            C.append(B[j])
            j += 1
    
    # Case 2 :- If list B is over, shift all elements of A to C
    while i < m:
        C.append(A[i])
        i += 1
    
    #Case 3 :- If list A is over, shift all elements of B to C 
    while j < n:
        C.append(B[j])
        j += 1
        
    return C

def merge_sort(L):
    n = len(L)
    if n<= 1:
        return L 

    left = merge_sort(L[:n//2])
    right = merge_sort(L[n//2:])
    
    sorted_merge = merge(left, right)
    return sorted_merge
    
```

```python
merge_sort([9, 1, 4, 9, 5])
```
