### Adjacency Matrix

```python
def get_adj_matrix(V, E):
    size = len(V)
    import numpy as np
    adj_matrix = np.zeros(shape=(size, size), dtype=int)
    for (u, v) in E:
        adj_matrix[u][v] = 1
    return adj_matrix
```

### Adjacency List

```python
def get_adj_list(V, E):
    size = len(V)
    adj_list = {i: [] for i in range(size)}
    for (u, v) in E:
        adj_list[u].append(v)
    return adj_list
```

### Graph Representation

```python
V = [0, 1, 2, 3, 4]
E = [(0, 1), (0, 2), (1, 3), (1, 4), (2, 4), (2, 3), (3, 4)]

print("Adjacency Matrix:")
print(get_adj_matrix(V, E))

print("\nAdjacency List:")
print(get_adj_list(V, E))
```
