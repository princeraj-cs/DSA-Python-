### BFS - Queue

```python
from collections import deque
```

```python
def bfs(AList, start_vertex):
    visited = {}
    
    for i in AList.keys():
        visited[i] = False
    # print("Before:", visited)
    
    q = deque()
    
    visited[start_vertex] = True 
    q.append(start_vertex)
    
    while (not len(q) == 0):
        current_vertex = q.popleft()
        for adj_vertex in AList[current_vertex]:
            if not visited[adj_vertex]:
                visited[adj_vertex] = True
                q.append(adj_vertex)

    return visited

AList = {0: [1, 2], 1: [3, 4], 2: [4, 3], 3: [4], 4: []}

print("After:",bfs(AList, 0))
```

### DFS - Stack

```python
def dfs(AList, start_vertex):
    visited = {}
    
    for i in AList.keys():
        visited[i] = False 

    st = deque()
    st.append(start_vertex)
    
    while(not len(st) == 0):
        current_vertex = st.pop()
        visited[current_vertex] = True 
        
        for adj_vertex in AList[current_vertex]:
            if not visited[adj_vertex]:
                st.append(adj_vertex)
    return visited

AList = {0: [1, 2], 1: [3, 4], 2: [4, 3], 3: [4], 4: []}
print(dfs(AList, 0))
```

### Find the Parent and level information

```python
def bfs_parent(AList, start_vertex):
    level, parent = {}, {}
    
    for i in AList.keys():
        level[i] = -1
        parent[i] = -1
        
    q = deque()
    level[start_vertex] = 0
    q.append(start_vertex)
    
    while (not len(q) == 0):
        current_vertex = q.popleft()
        for adj_vertex in AList[current_vertex]:
            if level[adj_vertex] == -1:
                level[adj_vertex] = level[current_vertex] + 1
                parent[adj_vertex] = current_vertex
                q.append(adj_vertex)
                
    return level, parent

AList = {0: [1, 2], 1: [3, 4], 2: [4, 3], 3: [4], 4: []}
print(bfs_parent(AList, 0))
```

### Connected Components

```python
def components(Alist):
    component = {}
    compid, seen = 0, 0
    
    for v in Alist.keys():
        component[v] = -1
    
    while (seen < max(Alist.keys())):
        startv = min(i for i in Alist.keys() if component[i] == -1)
        visited = bfs(Alist, startv)
        
        for v in visited.keys():
            if visited[v]:
                seen += 1
                component[v] = compid
        compid += 1
    return component 

AList = {0: [1], 1: [2], 2: [0], 3: [4, 6], 4: [3, 7], 5: [3, 7], 6: [5], 7: [4, 8], 8: [5, 9], 9: [8]}
components(AList)
    
    
```

### DFS List Pre Post - Recursion

```python
# Global variables
(visited,pre,post) = ({},{},{})

# Initialization
def DFSInitPrePost(AList):   
    for each_vertex in AList.keys():
        visited[each_vertex] = False
        (pre[each_vertex],post[each_vertex]) = (-1,-1)
    return

# Implementation of Pre and Post numbering on graph using DFS approch
def DFSListPrePost(AList,v,count):
    # Mark veretx v as visited
    visited[v] = True
    # Assigned the pre numbering for v when we are traversing the vertex(entering the vertex)
    pre[v] = count
    # Increment the count
    count = count + 1                      
    # Repeat for each unvisited adjacent of vertex v
    for adj_vertex in AList[v]:
        if (not visited[adj_vertex]):
            # Recursively call the DFSListPrePost on unvisited adjacent of v
            count = DFSListPrePost(AList,adj_vertex,count)
    # Assigned the post numbering for v when return back from recusrive call(leaving the vertex)
    post[v] = count
    # Increment the count
    count = count + 1                           
    
    return(count)


AList = {0: [1, 4],1: [0],2: [],3: [],4: [0, 8, 9],5: [],6: [],7: [],8: [4, 9],9: [8, 4]}
DFSInitPrePost(AList)
print(DFSListPrePost(AList,0,0))
print("Visited",visited)
print("Pre: ",pre)
print("Post" ,post)
```

```python
# Initialization Function
def DFSInitList(AList):
    (visited,parent) = ({},{})
    for each_vertex in AList.keys():
        visited[each_vertex] = False
        parent[each_vertex] = -1
    return(visited,parent)

# DFS Recursive Implementation for Adjacency list
def DFSList(AList,visited,parent,v):
    # Mark vertex v as visited vertex
    visited[v] = True
    # Repeat following for each unvisited adjacent of vertex v
    for adj_vertex in AList[v]:
        if (not visited[adj_vertex]):
            # Assign vertex v as parent of each unvisited adjacent of v 
            parent[adj_vertex] = v
            
            # Recursively call the DFS on unvisited adjacent of v
            (visited,parent) = DFSList(AList,visited,parent,adj_vertex)
            
    return(visited,parent)


AList ={0: [1, 2], 1: [3, 4], 2: [4, 3], 3: [4], 4: []}
v,p = DFSInitList(AList)
print(DFSList(AList,v,p,0))
```

### DFS Recursion

```python
def dfs(AList, start_vertex, result, visited):
    visited[start_vertex] = True
    result.append(start_vertex)
    
    for adj_vertex in AList[start_vertex]:
        if not visited[adj_vertex]:
            dfs(AList, adj_vertex, result, visited)
    
    return result

AList = {0: [1, 2], 1: [3, 4], 2: [4, 3], 3: [4], 4: []} 
visited = {vertex: False for vertex in AList.keys()}
result = []
print(dfs(AList, 0, result, visited))
```
