### Topolocial Sorting

```python
from collections import deque
```

```python
def topologial_sort(Alist):
    indegree, toposort = {}, []
    zdq = deque()
    
    for v in Alist.keys():
        indegree[v] = 0
    
    for v in Alist.keys():
        for u in Alist[v]:
            indegree[u] += 1 
    
    for u in Alist.keys():
        if indegree[u] == 0:
            zdq.append(u)
    
    while (not len(zdq) == 0):
        curr_vertex = zdq.popleft()
        toposort.append(curr_vertex)
        
        indegree[curr_vertex] -= 1
        
        for adj_vertex in Alist[curr_vertex]:
            indegree[adj_vertex] -= 1
            if indegree[adj_vertex] == 0:
                zdq.append(adj_vertex)
    return toposort


AList={0: [2, 3, 4], 1: [2, 7], 2: [5], 3: [5, 7], 4: [7], 5: [6], 6: [7], 7: []}
print(topologial_sort(AList))

            
```

### Longest Path

```python
def topologial_sort(Alist):
    indegree, lpath,toposort = {}, {},[]
    zdq = deque()
    
    for v in Alist.keys():
        indegree[v] = 0
        lpath[v] = 0
    
    for v in Alist.keys():
        for u in Alist[v]:
            indegree[u] += 1 
    
    for u in Alist.keys():
        if indegree[u] == 0:
            zdq.append(u)
    
    while (not len(zdq) == 0):
        curr_vertex = zdq.popleft()
        toposort.append(curr_vertex)
        
        indegree[curr_vertex] -= 1
        
        for adj_vertex in Alist[curr_vertex]:
            indegree[adj_vertex] -= 1
            lpath[adj_vertex] = max(lpath[adj_vertex], lpath[curr_vertex]+1)
            
            if indegree[adj_vertex] == 0:
                zdq.append(adj_vertex)
    return toposort, lpath


AList={0: [2, 3, 4], 1: [2, 7], 2: [5], 3: [5, 7], 4: [7], 5: [6], 6: [7], 7: []}
print(topologial_sort(AList))

            
```
