### **Quick Sort**

```python
def partition(L, lower, upper):
    # Select first element as a pivot
    pivot = L[lower]
    i = lower
    
    for j in range(lower+1, upper+1):
        if L[j] <= pivot:
            i = i+1
            L[i], L[j] = L[j], L[i]
        L[lower], L[i] = L[i], L[lower]
    return i # Return the position of pivot

def quik_sort(L, lower, upper):
    if(lower < upper):
        pivot_pos = partition(L, lower, upper)
        
        quik_sort(L, lower, pivot_pos-1) # Call the quick sort on the leftside part of pivot
        quik_sort(L, pivot_pos+1, upper) # Call the quick sort on the rightside part of pivot
    
    return L 
```

```python
L = [9, 4, 3, 10, 5, 12, 1, 7, 6]
quik_sort(L, 0, len(L) - 1)
```
