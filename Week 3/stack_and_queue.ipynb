### **Stack**

- Last in first out (LIFO)
- Reverse the string
- Evaluate Expression
- Undo/Redo Operation
- Backtracking
- Depth First Search(DFS) in Graph

```python
class Stack:
    def __init__(self):
        self.stack = []
    def is_empty(self):
        return (self.stack == [])
    def Push(self, v):
        self.stack.append(v)
    def Pop(self):
        v = None
        if not self.is_empty():
            v = self.stack.pop()
        return v 
    def __str__(self):
        print(str(self.stack))
        
S = Stack()
S.Push(20)
S.Push(10)
S.Push(20)
S.Push(40)
S.Push(50)
S.Pop()
S.Pop()
```

### **Queue**

```python
class Queue:
    def __init__(self):
       self.queue = []
    def is_empty(self):
        return (self.queue == []) 
    def enqueue(self, v):
        self.queue.append(v)
    def dequeue(self):
        if not self.is_empty():
            v = self.queue[0]
            self.queue = self.queue[1:]
        return v 
    def __str__(self):
            print(str(self.queue))        
```

```python
Q = Queue()
Q.enqueue(10)
Q.enqueue(20)
Q.enqueue(30)
Q.enqueue(40)
print(Q.dequeue())
```
