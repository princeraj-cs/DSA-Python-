### **Computing GCD**
- gcd(m, n) <= min(m, n)
- compute list of common factors from 1 to min(m, n)
- return the last such common factor

```python
def gcd(m, n):
    cf = []
    for i in range(1, min(m, n) + 1):
        if (m % i == 0) and (n % i == 0):
            cf.append(i)
    return cf[-1]
```

```python
gcd(4, 12)
```

```python
def gcd(m, n):
    for i in range(1, min(m, n) + 1):
        if (m % i == 0) and (n % i == 0):
            cf = i
    return cf
```

### Compute Prime

```python
def factors(n):
    f = []
    for i in range(1, n+1):
        if (n % i == 0):
            f.append(i)
    return f
```

```python
factors(20)
```

```python
def prime(n):
    return (factors(n) == [1, n])
```

```python
prime(5)
```

### GCD

```python
def gcd(m, n):
    a, b = max(m, n), min(m, n)
    
    if a % b == 0:
        return b
    else:
        return gcd(b, a-b)
```

```python
gcd(654488, 98696)
```
